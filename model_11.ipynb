```python
import sys
!{sys.executable} -m pip install scapy

from scapy.all import sniff, get_if_list

# List available interfaces to help the user choose correctly
print("Available interfaces:", get_if_list())

interface = input("Enter network interface (e.g., eth0): ")
packet_count = int(input("Enter number of packets to capture: "))

def display_packet(packet):
    print(packet.summary())

print(f"\nStarting capture on {interface}...")
try:
    sniff(iface=interface, count=packet_count, prn=display_packet, store=False)
except Exception as e:
    print(f"Error: {e}")
```
