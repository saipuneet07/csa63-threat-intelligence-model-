```python
logs = [
    "2026-08-25 INFO user=alice action=login status=success",
    "2026-08-25 ERROR user=bob action=login status=failed",
    "2026-08-25 INFO user=charlie action=file_download status=success",
    "2026-08-25 ERROR user=alice action=payment status=failed"
]

index = []

for line in logs:
    parts = line.split()
    event = {
        "date": parts[0],
        "level": parts[1],
        "user": parts[2].split("=")[1],
        "action": parts[3].split("=")[1],
        "status": parts[4].split("=")[1]
    }
    index.append(event)

print("Structured events:")
for event in index:
    print(event)

print("\nError count by user:")
counts = {}

for event in index:
    if event["level"] == "ERROR":
        user = event["user"]
        counts[user] = counts.get(user, 0) + 1

for user, count in counts.items():
    print(user, count)
```
