```python
policy = input("Enter draft security policy: ").lower()

required_elements = {
    "scope": ["scope", "applies to", "coverage"],
    "roles and responsibilities": ["roles", "responsibilities", "duties"],
    "enforcement": ["enforcement", "penalty", "violation", "disciplinary"],
    "review cycle": ["review", "annually", "quarterly", "cycle"],
    "approval authority": ["approval", "approved by", "authority", "management"]
}

missing = []

for element, keywords in required_elements.items():
    found = any(keyword in policy for keyword in keywords)

    if not found:
        missing.append(element)

if missing:
    print("Missing governance elements:")
    for element in missing:
        print(element)
else:
    print("All required governance elements are present")
```
