```python
connections = set()

packets = [
    ("outbound", "192.168.1.10", "8.8.8.8", 5000, 53),
    ("inbound", "8.8.8.8", "192.168.1.10", 53, 5000),
    ("inbound", "10.0.0.5", "192.168.1.10", 4444, 22),
    ("outbound", "192.168.1.10", "example.com", 5001, 80),
    ("inbound", "example.com", "192.168.1.10", 80, 5001)
]

for direction, source, destination, source_port, destination_port in packets:
    if direction == "outbound":
        connections.add((destination, source, destination_port, source_port))
        print("Allowed outbound:", source, "->", destination)

    elif (source, destination, source_port, destination_port) in connections:
        print("Allowed return traffic:", source, "->", destination)

    else:
        print("Blocked unsolicited inbound:", source, "->", destination)
```
