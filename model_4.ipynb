```python
import hashlib

def generate_stream(key, length):
    stream = b""
    counter = 0

    while len(stream) < length:
        data = key + str(counter)
        stream += hashlib.sha256(data.encode()).digest()
        counter += 1

    return stream[:length]

def encrypt(message, key):
    message_bytes = message.encode()
    stream = generate_stream(key, len(message_bytes))
    return bytes(m ^ s for m, s in zip(message_bytes, stream))

def decrypt(ciphertext, key):
    stream = generate_stream(key, len(ciphertext))
    decrypted = bytes(c ^ s for c, s in zip(ciphertext, stream))

    try:
        return decrypted.decode()
    except UnicodeDecodeError:
        return str(decrypted)

message = input("Enter data to send: ")
shared_key = input("Enter shared key: ")
attacker_key = input("Enter attacker key: ")

ciphertext = encrypt(message, shared_key)

print("Captured data in transit:", ciphertext.hex())
print("Attacker output:", decrypt(ciphertext, attacker_key))
print("Receiver output:", decrypt(ciphertext, shared_key))
```
