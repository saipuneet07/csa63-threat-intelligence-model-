```python
import re
from urllib.parse import urlparse

url = input("Enter URL: ").lower()
parsed = urlparse(url if "://" in url else "http://" + url)
domain = parsed.netloc

suspicious_keywords = ["login", "verify", "secure", "update", "account", "banking", "password", "free", "bonus"]
indicators = []

if any(keyword in url for keyword in suspicious_keywords):
    indicators.append("Suspicious keyword found")

if re.match(r"^\d{1,3}(\.\d{1,3}){3}$", domain):
    indicators.append("IP-based domain found")

if domain.count(".") > 2:
    indicators.append("Excessive subdomains found")

if "@" in url:
    indicators.append("@ symbol found in URL")

if "-" in domain:
    indicators.append("Hyphen found in domain")

if indicators:
    print("Possible phishing URL")
    print("Indicators:")
    for indicator in indicators:
        print(indicator)
else:
    print("URL seems safe")
```
