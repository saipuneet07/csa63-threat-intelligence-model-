```python
alert = input("Enter SOC alert description: ").lower()

p1_keywords = ["ransomware", "data breach", "domain admin", "critical server", "active exfiltration"]
p2_keywords = ["malware", "privilege escalation", "multiple systems", "suspicious login"]
p3_keywords = ["failed login", "policy violation", "single host", "blocked attack"]
p4_keywords = ["low risk", "informational", "scan", "test alert"]

if any(keyword in alert for keyword in p1_keywords):
    priority = "P1"
elif any(keyword in alert for keyword in p2_keywords):
    priority = "P2"
elif any(keyword in alert for keyword in p3_keywords):
    priority = "P3"
elif any(keyword in alert for keyword in p4_keywords):
    priority = "P4"
else:
    priority = "P4"

print("SOC alert priority:", priority)
```
