```python
access_policy = {
    "guest": ["internet"],
    "employee": ["internet", "intranet"],
    "finance": ["internet", "intranet", "finance_server"],
    "admin": ["internet", "intranet", "finance_server", "domain_controller"]
}

source = input("Enter compromised segment: ").lower()
destination = input("Enter target segment: ").lower()

if source in access_policy and destination in access_policy[source]:
    print("Access allowed")
    print("Lateral movement possible")
else:
    print("Access blocked")
    print("Network segmentation stopped lateral movement")
```
