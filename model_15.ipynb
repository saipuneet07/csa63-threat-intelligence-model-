```python
case_study = input("Enter incident case study: ").lower()

categories = {
    "Phishing": ["phishing", "email", "fake link", "credential", "login page", "attachment"],
    "Unpatched vulnerability": ["unpatched", "vulnerability", "exploit", "outdated", "patch", "cve"],
    "Misconfiguration": ["misconfiguration", "open storage", "default password", "public access", "wrong setting"],
    "Insider threat": ["employee", "insider", "staff", "authorized user", "data theft", "privilege misuse"],
    "Supply-chain compromise": ["vendor", "third party", "supplier", "software update", "dependency", "supply chain"]
}

scores = {}

for category, keywords in categories.items():
    scores[category] = sum(1 for keyword in keywords if keyword in case_study)

result = max(scores, key=scores.get)

if scores[result] == 0:
    result = "Unknown"

print("Root cause category:", result)
```
