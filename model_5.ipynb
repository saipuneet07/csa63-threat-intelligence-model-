```python
description = input("Enter threat-intelligence report description: ").lower()
audience = input("Enter intended audience: ").lower()

text = description + " " + audience

categories = {
    "Strategic": ["executive", "management", "business", "risk", "trend", "industry", "long-term"],
    "Tactical": ["analyst", "tactic", "technique", "procedure", "ttp", "malware", "campaign"],
    "Operational": ["soc", "incident", "ip address", "domain", "hash", "indicator", "alert"]
}

scores = {}

for category, keywords in categories.items():
    scores[category] = sum(1 for keyword in keywords if keyword in text)

result = max(scores, key=scores.get)

if scores[result] == 0:
    result = "Unknown"

print("Report classification:", result)
```
