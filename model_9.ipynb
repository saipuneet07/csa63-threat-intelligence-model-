```python
activity = input("Enter intelligence team activity: ").lower()

stages = {
    "Direction": ["requirement", "goal", "objective", "planning", "priority"],
    "Collection": ["collect", "gather", "source", "feed", "log"],
    "Processing": ["clean", "normalize", "filter", "format", "organize"],
    "Analysis": ["analyze", "pattern", "correlate", "assess", "conclude"],
    "Dissemination": ["share", "report", "brief", "distribute", "deliver"],
    "Feedback": ["feedback", "review", "improve", "evaluate", "response"]
}

scores = {}

for stage, keywords in stages.items():
    scores[stage] = sum(1 for keyword in keywords if keyword in activity)
result = max(scores, key=scores.get)

if scores[result] == 0:
    result = "Unknown"

print("Threat Intelligence Lifecycle stage:", result)
```
